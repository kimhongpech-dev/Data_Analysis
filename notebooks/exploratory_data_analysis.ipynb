# Exploratory data analysis

Use this notebook for initial exploration before building or extending the dashboard. The cleaning helpers are shared with `src/clean_data.py`.

Run the cells from top to bottom.

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT / "src"))

import pandas as pd

from clean_data import clean_dataframe, read_raw_file

RAW_DIR = PROJECT_ROOT / "data" / "raw"
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"

print(f"Project root: {PROJECT_ROOT}")
```

## 1. List the raw files

```python
raw_files = sorted(
    path
    for path in RAW_DIR.iterdir()
    if path.suffix.lower() in {".csv", ".xlsx", ".xlsm", ".json"}
)
raw_files
```

## 2. Load and inspect one file

Change the index if you have more than one raw file.

```python
if not raw_files:
    raise FileNotFoundError(
        "Add a CSV, XLSX, XLSM, or JSON file to data/raw/ first."
    )

path = raw_files[0]  # change this index to explore another file
df = read_raw_file(path)

print(f"{path.name}: {len(df):,} rows x {df.shape[1]} columns")
df.head()
```

## 3. Clean the data

```python
clean_df, report = clean_dataframe(df, name=path.stem, fill_missing=False)
report
```

## 4. Missing values

```python
missing = clean_df.isna().sum().sort_values(ascending=False)
missing[missing > 0].to_frame("missing")
```

## 5. Summary statistics

```python
clean_df.describe(include="all").T
```

## 6. Save the cleaned data

This produces the same kind of file the dashboard reads.

```python
output_path = PROCESSED_DIR / f"{path.stem}_clean.csv"
clean_df.to_csv(output_path, index=False, encoding="utf-8-sig")
print(f"Saved {output_path.relative_to(PROJECT_ROOT)}")
```
